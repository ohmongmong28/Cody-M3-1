# AI 반도체 시계열 분석: 엔비디아(NVDA) vs TSMC(TSM)
본 노트북은 `run_analysis.py`와 동일한 분석 파이프라인을 단계별로 재현합니다.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

df = pd.read_csv('data/ai_chip_merged.csv', parse_dates=['Date'])
print(f'데이터 기간: {df["Date"].min().date()} ~ {df["Date"].max().date()}, 총 {len(df)}거래일')
df.head()
```

```
# 기초 통계량 및 일간 수익률 상관계수 확인
display(df[['NVDA_Close', 'TSM_Close', 'NVDA_Ret', 'TSM_Ret', 'NVDA_Vol20', 'TSM_Vol20']].describe().round(2))
print('일간 수익률 상관계수:', round(df['NVDA_Ret'].corr(df['TSM_Ret']), 4))
```

```
# 시계열 분해(보너스 심화) 검증 코드
decomp = seasonal_decompose(df.set_index('Date')['NVDA_Close'], model='additive', period=63)
fig = decomp.plot()
fig.set_size_inches(11, 7)
plt.tight_layout()
plt.show()
```
